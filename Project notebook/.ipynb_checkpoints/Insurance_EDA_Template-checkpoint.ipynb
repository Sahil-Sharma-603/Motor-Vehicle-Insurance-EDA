# Motor Vehicle Insurance - Exploratory Data Analysis

## Project Overview
This notebook explores motor vehicle insurance data to understand customer behavior, policy characteristics, and claims patterns.

**Dataset**: Motor Vehicle Insurance Data from Mendeley  
**Rows**: ~105,000 policy records  
**Objective**: Discover insights about insurance premiums, claims, and customer demographics

## Step 1: Import Libraries

We'll use three main libraries:
- **pandas**: For data manipulation
- **matplotlib**: For creating visualizations
- **seaborn**: For beautiful statistical plots

```python
# Import pandas, matplotlib.pyplot, and seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style and figure size
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,6)
    
# Set pandas display options
pd.set_option('display.max_columns', None)

print("All libraries successfully imported")
```

## Step 2: Load the Data

The data is in CSV format with semicolon (;) as the separator.

```python
# Load the CSV file with sep=';'

df = pd.read_csv('/Users/Sahil/Documents/Data Analytic Project/Data Analytics Project by Harsh/Project 1/Dataset of an actual motor vehicle insurance portfolio/Motor vehicle insurance data.csv', sep = ';')

# Print the shape of the dataset
print(f"Dataset Loaded : {df.shape[0]:,} rows and {df.shape[1]} columns")

# Display first 5 rows
print("\nFirst 5 rows:")
df.head()
```

## Step 3: Understand the Data Structure

Let's examine what columns we have and their data types.

```python
# Use .info() to check data types
df.info()
```

```python
# Print all column names
print("Column Names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}:{col}")
```

## Step 4: Check for Missing Values

Missing data can affect our analysis. Let's see where we have gaps.

```python
# Calculate missing values count and percentage

missing = df.isnull().sum()
missing_percentage = (missing/len(df))*100


# Create a summary DataFrame
missing_summary = pd.DataFrame({
    'Missing_Count': missing,
    'Percentage': missing_percentage })


# Display columns with missing values
missing_summary = missing_summary[missing_summary['Missing_Count']>0].sort_values('Missing_Count', ascending=False)

print("Columns with missing values:")
missing_summary
```

## Step 5: Summary Statistics

Let's look at numerical summaries to understand the range and distribution of our data.

```python
# Use .describe() to get summary statistics
df.describe()
```

### Key Observations from Summary:
- **Premium**: Average premium is around €316
- **Cost_claims_year**: Most policies have no claims (mean close to 0)
- **Vehicle age**: Cars range from very old to brand new
- **Vehicle Value**: Ranges from scrap value (270) to luxury vehicle (220K)

## Step 6: Analyze Premium Distribution

**Question**: How are insurance premiums distributed?

```python
# Create a figure with 2 subplots
plt.figure(figsize = (12,5))

# Plot histogram of Premium in first subplot
plt.subplot(1,2,1)
plt.hist(df['Premium'], bins=50, color='purple', edgecolor='black', alpha=0.7)
plt.xlabel('Premiums(Euros)')
plt.ylabel('Frequency')
plt.title('Distribution of Insurance Premiums')

# Add mean line to histogram
plt.axvline(df['Premium'].mean(), color='red', linestyle='--', label=f'Mean: Euro {df["Premium"].mean():.2f}')
plt.legend()


# Plot boxplot of Premium in second subplot
plt.subplot(1,2,2)
plt.boxplot(df['Premium'], orientation='vertical')
plt.ylabel('Premiums(Euros)')
plt.title('Premium Box Plot')
plt.tight_layout()
plt.show()


# Print summary statistics
print(f"Average Premium: €{df['Premium'].mean():.2f}")
print(f"Median Premium: €{df['Premium'].median():.2f}")
print(f"Min Premium: €{df['Premium'].min():.2f}")
print(f"Max Premium: €{df['Premium'].max():.2f}")

```

## Step 7: Claims Analysis

**Question**: How many policies have claims?

```python
# Count policies with claims (N_claims_year > 0)
claims_count = (df['N_claims_year']>0).sum()

# Count policies without claims
no_claims_count = (df['N_claims_year']==0).sum()

# Print counts and percentages
print(f"Policies with claims: {claims_count:,} ({claims_count/len(df)*100:.1f}%)")
print(f"Policies without claims: {no_claims_count:,} ({no_claims_count/len(df)*100:.1f}%)")


# Create pie chart 
plt.figure(figsize = (8,6))
labels = ['No Claims', 'Has Claims']
sizes = [no_claims_count, claims_count]
colors = ['lightgreen', 'salmon']
explode = (0.05,0)

plt.pie(sizes, explode = explode, labels = labels, colors = colors, autopct = '%1.1f%%', 
        shadow=True, startangle=90, textprops={'fontsize':12, 'weight':'bold'})

plt.show()


```

## Step 8: Premium vs Claims Cost

**Question**: Is there a relationship between premium and claims cost?

```python
# Filter to policies with claims
df_with_claims = df[df['Cost_claims_year']>0]
print(f"Analyzing {len(df_with_claims):,} policies that had claims...\n")

# Create scatter plot of Premium vs Cost_claims_year
plt.figure(figsize = (10,6))
plt.scatter(df_with_claims['Premium'], df_with_claims['Cost_claims_year'],
            alpha=0.5, color='purple', edgecolor='black', linewidth=0.5 )
plt.xlabel('Premium (€)', fontsize=12)
plt.ylabel('Claim cost (€)', fontsize=12)
plt.title('Premium vs Claims Cost (Policies with Claim Only)', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3)
plt.show()

# Calculate correlation
correlation = df_with_claims['Premium'].corr(df_with_claims['Cost_claims_year'])
print(f"Correlation between Premium and Claims Cost: {correlation:.3f}")
```

## Step 9: Vehicle Age Analysis

**Question**: How old are the insured vehicles?

```python
# Calculate vehicle age (2020 - Year_matriculation)
df['Vehicle_Age'] = 2020-df['Year_matriculation']


# Filter to valid ages (>= 0)
df_valid_age = df[df['Vehicle_Age'] >= 0]

# Plot histogram of vehicle ages
plt.subplot(1,2,1)
plt.hist(df_valid_age['Vehicle_Age'], bins=30, color='purple', edgecolor='black', alpha=0.7)
plt.xlabel('Vechile Age (Years)')
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Vehicle Ages',fontsize=14, weight='bold')

# Add mean line to histogram
plt.axvline(df_valid_age['Vehicle_Age'].mean(), color='red', linestyle='--', label=f'Mean: {df_valid_age['Vehicle_Age'].mean():.1f}')
plt.legend()


# Create age categories and plot bar chart
plt.subplot(1,2,2)
age_bins = [0,3,7,10,15,25,100]
age_labels = ['New (0-3)', 'Young (4-7)', 'Younger Mid (8-10)', 'Mid (10-15)', 'Older Mid (15-25)', 'Old (25+)' ]
df_valid_age['Age_Category'] = pd.cut(df_valid_age['Vehicle_Age'], bins=age_bins, labels=age_labels)

age_counts = df_valid_age['Age_Category'].value_counts().sort_index()

age_counts.plot(kind='bar', color='purple', alpha=0.7, edgecolor='black')
plt.xlabel('Vehicle Age Category', fontsize=12)
plt.ylabel('Number of Policies', fontsize=12)
plt.title('Policies by Vehicle Age Category', fontsize=14, weight='bold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()



# Print average and median age
print(f"Average Vehicle Age: {df_valid_age['Vehicle_Age'].mean():.1f} years")
print(f"Median Vehicle Age: {df_valid_age['Vehicle_Age'].median():.1f} years")
```

## Step 10: Fuel Type Analysis

**Question**: What fuel types are most common?

```python
# Count fuel types using value_counts()
fuel_counts = df['Type_fuel'].value_counts()
print('Fuel Type Distribution')
print(fuel_counts)

# Create bar chart
plt.figure(figsize = (10,6))
color = ['purple','lightblue']
fuel_counts.plot(kind='bar', color=color, alpha=0.7, edgecolor='black')
plt.xlabel('Fuel Type',fontsize=12)
plt.ylabel('Number of Policies', fontsize=12)
plt.title('Distribution of Fuel Types', fontsize=14, weight='bold')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)


# Add percentage labels
for i, v in enumerate(fuel_counts):
    plt.text(i,v + 1000, f'{v/len(df)*100: .1f}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()

```

## Step 11: Premium by Vehicle Age

**Question**: Do older vehicles have different premiums?

```python
# Group by age category and calculate mean/median premium
premium_by_age = df_valid_age.groupby('Age_Category')['Premium'].agg(['mean','median','count'])
premium_by_age.columns = ['Average Premium','Median Premium','Policy Count']

print("Premium Analysis by Vehicle Age:")
print(premium_by_age)
print()


#Create bar chart with average and median
plt.figure(figsize = (10,6))
x = range(len(premium_by_age))
plt.bar(x, premium_by_age['Average Premium'], color='purple', alpha=0.7, edgecolor='black', label = 'Average')
plt.plot(x, premium_by_age['Median Premium'], color='red', marker='o', linewidth=2, markersize=8, label = 'Median')

plt.xlabel('Vehicle Age Category',fontsize=12)
plt.ylabel('Premium (€)', fontsize=12)
plt.title('Insurance Premium by Vehicle Age', fontsize=14, weight='bold')
plt.xticks(x,premium_by_age.index,rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()


```

## Step 12: Claims Rate by Vehicle Age

**Question**: Do older vehicles have more claims?

```python
# Calculate total claims and policies by age category
claims_by_age = df_valid_age.groupby('Age_Category').agg({
    'N_claims_year':'sum',
    'ID':'count'

})

claims_by_age.columns = ['Total_Claims','Total_Policies']

# Calculate claims rate percentage
claims_by_age['Claims_Rate'] = (claims_by_age['Total_Claims']/claims_by_age['Total_Policies'])*100
print("Claims Analysis by Vehicle Age:")
print(claims_by_age)
print()


# Create bar chart with value labels
plt.figure(figsize = (10,6))
x = range(len(claims_by_age))
plt.bar(x, claims_by_age['Claims_Rate'], color='Salmon', alpha=0.7, edgecolor='black')
plt.xlabel('Vehicle Age Category',fontsize=12)
plt.ylabel('Claims Rate (%)', fontsize=12)
plt.title('Claims Rate by Vehicle Age', fontsize=14, weight='bold')
plt.xticks(x,claims_by_age.index,rotation=45)
plt.grid(axis='y', alpha=0.3)


# Add value labels
for i, v in enumerate(claims_by_age['Claims_Rate']):
    plt.text(i,v + 0.5, f'{v:.1f}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Step 13: Vehicle Power Analysis

**Question**: How does vehicle power affect premiums?

```python
# Create power categories
power_bins = [0, 75, 100, 150, 500]
power_labels = ['Low (0-75)', 'Medium (76-100)', 'High (101-150)', 'Very High (150+)']
df['Power_Category'] = pd.cut(df['Power'], bins=power_bins, labels=power_labels)

# Calculate average premium by power category
preimium_by_power = df.groupby('Power_Category')['Premium'].agg(['mean','count'])
preimium_by_power.columns = ['Average Premium', 'Policy Count']

print("Premium by Vehicle Power:")
print(preimium_by_power)
print()

# Create bar chart
plt.figure(figsize = (10,6))
preimium_by_power['Average Premium'].plot(kind='bar', color='purple', alpha=0.7, edgecolor='black')
plt.xlabel('Vehicle Power Category',fontsize=12)
plt.ylabel('Average Premium (€)', fontsize=12)
plt.title('Average Premium by Vehicle Power', fontsize=14, weight='bold')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

# Add value labels
for i, v in enumerate(preimium_by_power['Average Premium']):
    plt.text(i,v + 5, f'€{v:.0f}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()



```

## Step 14: Correlation Heatmap

**Question**: Which numerical variables are related to each other?

```python
# Select numerical columns
numerical_cols = ['Premium', 'Cost_claims_year', 'N_claims_year',
                  'Power', 'Cylinder_capacity', 'Value_vehicle', 'Vehicle_Age']

# Calculate correlation matrix
correlation_matrix = df_valid_age[numerical_cols].corr()

# Create heatmap using seaborn
plt.figure(figsize = (10,8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='RdPu',
            square=True, linewidth=1, cbar_kws={"shrink":0.8})
plt.title('Corrleation Heatmap of Numerical Variables', fontsize=14, weight='bold', pad=20)
plt.tight_layout()
plt.show()


# Find strongest correlations
print("\n Strongest Positive Correlations:")
corr_pairs = correlation_matrix.unstack()
corr_pairs = corr_pairs[corr_pairs < 1] # remove diagonal
print(corr_pairs.sort_values(ascending=False).head(5))
```
